## 1. Conditional Statements
Conditions let Python choose what to do based on `True` or `False`. End each condition header with `:` and indent its body consistently, usually with four spaces.

| Keyword | Purpose |
|---|---|
| `if` | Check the first condition |
| `elif` | Check another condition if earlier ones were false |
| `else` | Run a fallback if no earlier condition was true |

In one `if`/`elif`/`else` chain, only the first matching branch runs. Separate `if` statements can each run.

```python
scooby_is_hungry = True

if scooby_is_hungry:
    print('Shaggy gives Scooby a snack.')

print('The investigation continues.')
```

### `if`
`if` runs code when a condition is true. Set a default result first if needed. Use `==` to compare values and `=` to assign them.

```python
velma_clue = 'muddy footprint'
expected_clue = 'muddy footprint'
result = 'Velma needs another clue.'

if velma_clue == expected_clue:
    result = 'Velma found a matching clue!'

print(result)
```

### `elif`
`elif` checks another condition when earlier ones are false. The first match runs. Try clue counts `1`, `2`, `5`, and `6`.

```python
fred_clue_count = 3

if fred_clue_count < 2:
    investigation_stage = 'Fred needs more clues.'
elif fred_clue_count <= 5:
    investigation_stage = 'Fred is planning a trap.'
else:
    investigation_stage = 'Fred is ready to reveal the suspect.'

print(investigation_stage)
```

### `else`
`else` has no condition. It runs when all earlier conditions in the chain are false.

```python
daphne_has_key = False

if daphne_has_key:
    print('Daphne opens the locked room.')
else:
    print('Daphne searches for the key.')
```

## 2. Logical Operators
Combine or reverse conditions using `and`, `or`, and `not`.

| Operator | Meaning |
|---|---|
| `and` | Both conditions must be true |
| `or` | At least one condition must be true |
| `not` | Reverse the condition |

```python
scooby_has_snack = True
shaggy_has_flashlight = True
monster_nearby = False

if scooby_has_snack and shaggy_has_flashlight:
    print('Scooby and Shaggy are ready to explore.')

if scooby_has_snack or shaggy_has_flashlight:
    print('They have at least one useful supply.')

if not monster_nearby:
    print('The hallway is clear.')
```

### Combine comparisons and logic
Parentheses help make combined conditions readable. Change the values to try the other branch.

```python
velma_skill = 'clue analysis'
required_skill = 'clue analysis'
velma_clue_count = 4

if (velma_skill == required_skill) and (velma_clue_count >= 2):
    print('Velma has the skill and evidence to investigate.')
else:
    print('The gang needs more preparation.')
```

## 3. `pass`
`pass` does nothing. Use it as a placeholder when a code block is not ready. The rest of the program keeps running.

```python
fred_has_plan = True

if fred_has_plan:
    pass  # Add the trap-building steps later

print('The gang keeps investigating.')
```

## 4. Putting It Together
Use `if`, `elif`, and `else` to choose one next step. Check the skill first, then the clue count.

```python
detective = {
    'name': 'Velma',
    'skill': 'clue analysis',
    'clue_count': 3,
}
required_skill = 'clue analysis'

if detective['skill'] != required_skill:
    next_step = 'Ask another member of the gang for help.'
elif detective['clue_count'] < 2:
    next_step = 'Collect more clues.'
else:
    next_step = 'Explain the evidence to Mystery Inc.'

print(detective['name'] + ': ' + next_step)
```

## 5. Debugging
Common mistakes:
- **Assignment instead of comparison:** use `if velma_clue == expected_clue:`, not `if velma_clue = expected_clue:`.
- **Missing colon:** write `if scooby_is_hungry:`.
- **Missing indentation or an empty body:** indent a statement or use `pass`.
- **Misspelled variables:** `required_skil` and `required_skill` are different names; an undefined name raises `NameError`.

The cell below uses the corrected syntax.

```python
scooby_snacks = 2
required_snacks = 2

if scooby_snacks == required_snacks:
    print('Scooby has exactly enough snacks.')
else:
    print('Check Scooby’s snack supply.')
```

## 6. Lists
Lists store ordered items inside `[]`. They can contain duplicates and different types. You can change their items.

```python
gang = ['Scooby', 'Shaggy', 'Velma', 'Daphne', 'Fred']
print(gang)
print('Members:', len(gang))
```

### Indexing and changing items
Indexes start at `0`. Negative indexes count from the end; `-1` selects the last item.

```python
print(gang[0])
print(gang[-1])
gang[0] = 'Scooby-Doo'
print(gang)
```

### Adding and removing items
`append()` adds at the end. `insert(index, item)` adds at a position. `remove(value)` removes the first match and fails if it is missing. Put a comma between `insert()` arguments.

```python
clues = ['footprint', 'torn costume']
clues.append('secret passage')
clues.insert(1, 'muddy key')
clues.remove('torn costume')
print(clues)
print('Clue count:', len(clues))
```

### Combining lists
`+` makes a combined list. `extend()` adds each item from another list. `append(other_list)` adds the whole list as one nested item.

```python
velma_clues = ['footprint', 'muddy key']
daphne_clues = ['secret passage']
all_clues = velma_clues + daphne_clues
print(all_clues)

all_clues.extend(['costume receipt'])
print(all_clues)

fred_clues = []
for clue in all_clues:
    fred_clues.append(clue)
print(fred_clues)
```

### Joining strings
`separator.join(items)` turns a sequence of strings into one string. Every item must be a string.

```python
print('Mystery Inc.: ' + ', '.join(gang))
print('Evidence: ' + ' | '.join(all_clues))
```

### Slicing
Use `items[start:stop:step]`. `stop` is excluded. For a positive step, omitted bounds go from the beginning to the end. The default step is `1`. `[:]` makes a shallow copy: nested objects are still shared.

```python
print(gang[:2])   # First two members
print(gang[2:])   # From index 2 onward
print(gang[-2:])  # Last two members
print(gang[-1:])  # Last member as a list
print(gang[::2])  # Every second member
print(gang[:])    # Shallow copy
```

### Unpacking
Unpacking assigns list items to separate variables. Match the number of variables to the number of items. Use `*name` to collect the remaining items into a list.

```python
scooby, shaggy, velma, daphne, fred = gang
print(scooby, velma)

first_member, *other_members = gang
print(first_member)
print(other_members)
```

### Lists with conditions
Combine lists and conditional statements to choose the gang’s next step. `in` checks whether a value is present in a list.

```python
clues = ['footprint', 'muddy key', 'secret passage']

if 'muddy key' in clues and len(clues) >= 3:
    print('Velma has enough clues to check the locked room.')
elif len(clues) > 0:
    print('Daphne and Fred keep searching for evidence.')
else:
    print('Scooby and Shaggy start looking for clues.')
```

## 7. For Loops
A `for` loop repeats code for each item. Use it to go through a list or range. End the header with `:` and indent the code below it.

```python
gang = ['Scooby', 'Shaggy', 'Velma', 'Daphne', 'Fred']
for member in gang:
    print(member + ' joins the investigation.')
```

### Repeating with `range()`
`range(start, stop)` includes `start` and excludes `stop`. Here, Fred checks rooms 1, 2, and 3.

```python
for room_number in range(1, 4):
    print('Fred checks room', room_number)
```

### Conditions inside a loop
A loop can check a condition for each item. Here, Velma only reports the matching clue.

```python
clues = ['footprint', 'muddy key', 'secret passage']
for clue in clues:
    if clue == 'muddy key':
        print('Velma found the key to the locked room!')
```

## 8. While Loops
A `while` loop repeats while its condition is true. Change the value being checked so the loop can finish. If it starts false, the body is skipped.

```python
scooby_snacks = 3
while scooby_snacks > 0:
    print('Scooby eats a snack.')
    scooby_snacks -= 1

print('Snacks left:', scooby_snacks)
```

### Stop early with `break`
`break` stops the nearest loop immediately. This example also limits the number of attempts. Use `for` for a known sequence and `while` for a changing condition.

```python
attempts = 0
while attempts < 5:
    attempts += 1
    print('Shaggy searches cupboard', attempts)
    if attempts == 3:
        print('Shaggy finds the hidden clue!')
        break
```

## 9. String Formatting
String formatting puts values into readable messages. F-strings are a simple choice. The examples also show other styles you may see.

```python
character_name = 'Velma'
role = 'Detective'
clue_count = 4
```

### Output in Jupyter Notebook
Jupyter shows the last expression in a code cell. Strings may appear with quotes or escaped line breaks. Assigning a value alone shows no output.

```python
message = f'{character_name} found {clue_count} clues.'
message
```

`print()` shows text without surrounding quotes and displays line breaks. A plain expression in a `.py` script does not display automatically.

```python
print(message)
print('Mystery Inc. report:\nVelma found a clue.\nFred is planning a trap.')
```

### Concatenation with `+`
Join strings directly. Convert numbers with `str()` before concatenating them.

```python
'Detective: ' + character_name + ', clues: ' + str(clue_count)
```

### Percent formatting
An older style uses `%s` for strings and `%d` for integers. Multiple values go in a tuple.

```python
'Detective: %s, clues: %d' % (character_name, clue_count)
```

### F-strings
Prefix the string with `f` and place variables or expressions inside `{}`. Without the `f`, braces stay literal text.

```python
f'{character_name} is a {role.lower()} with {clue_count} clues.'
```

```python
f'If Daphne finds one more clue, the gang will have {clue_count + 1} clues.'
```

### `str.format()`
Put placeholders inside `{}` and pass values to `.format()`. You can use names for the placeholders.

```python
'Detective: {}, clues: {}'.format(character_name, clue_count)
```

```python
'{name} and {partner} investigate the mansion.'.format(name='Scooby', partner='Shaggy')
```

### Joining text with `join()`
A separator joins a sequence of strings. Every item must be a string. Passing a single string joins its individual characters.

```python
gang = ['Scooby', 'Shaggy', 'Velma', 'Daphne', 'Fred']
f'Mystery Inc.: {", ".join(gang)}' 
```

```python
' - '.join('CLUE')
```

### Formatting numbers
In an f-string, `:.2f` shows two decimal places, `:,` adds thousands separators, and `:.1%` shows a percentage. These change the text you see, not the original number.

```python
snack_price = 2.5
snack_budget = 1250
clues_found = 3
total_clues = 8
print(f'Scooby snack price: ${snack_price:.2f}')
print(f'Shaggy snack budget: ${snack_budget:,.2f}')
print(f'Investigation progress: {clues_found / total_clues:.1%}')
```

### A multiline notebook report
Use a triple-quoted f-string with `print()` for several lines. Plain Markdown cells do not replace `{variable}` with Python values.

```python
report = f'''Mystery Inc. Case Report
Detective: {character_name}
Clues found: {clue_count}
Team: {', '.join(gang)}
Snack price: ${snack_price:.2f}'''
print(report)
```

### Displaying multiple results in Jupyter
`display()` shows a result wherever you call it in a Jupyter cell. Use it to show several results in one cell.

```python
from IPython.display import display

display(f'Detective: {character_name}')
display(f'Clues: {clue_count}')
```
