## 1. Classes and Objects
A class is a blueprint. An object is one example made from that class. Attributes store its data. `__init__()` sets up a new object, and `self` refers to that object.

```python
class Pokemon:
    def __init__(self, name, pokemon_type, level):
        self.name = name
        self.pokemon_type = pokemon_type
        self.level = level

pikachu = Pokemon('Pikachu', 'Electric', 12)
bulbasaur = Pokemon('Bulbasaur', 'Grass/Poison', 10)
print(pikachu.name, pikachu.pokemon_type, pikachu.level)
print(bulbasaur.name, bulbasaur.pokemon_type, bulbasaur.level)
print(isinstance(pikachu, Pokemon))
```

## 2. Instance Methods
Methods are functions inside a class. They use `self` to read or change an object’s attributes. Call a method with `object.method()`.

```python
class TrainingPokemon:
    def __init__(self, name, level):
        self.name = name
        self.level = level

    def introduce(self):
        return f'{self.name} is level {self.level}.'

    def train(self):
        self.level += 1

charmander = TrainingPokemon('Charmander', 8)
print(charmander.introduce())
charmander.train()
print(charmander.introduce())
```

## 3. Class Variables
Instance variables belong to each object. Class variables are shared defaults. Update a shared counter through the class name. Here, the counter records how many objects have been created.

```python
class RegisteredPokemon:
    region = 'Kanto'
    total_registered = 0

    def __init__(self, name):
        self.name = name
        RegisteredPokemon.total_registered += 1

    def introduce(self):
        return f'{self.name} is registered in {self.region}.'

eevee = RegisteredPokemon('Eevee')
squirtle = RegisteredPokemon('Squirtle')
print(eevee.introduce())
print(squirtle.introduce())
print('Registered:', RegisteredPokemon.total_registered)
```

### Changing a shared default
Changing `ClassName.region` updates the shared default. Setting `object.region` gives that object its own value instead.

```python
RegisteredPokemon.region = 'Johto'
eevee.region = 'Kanto'
print(eevee.introduce())
print(squirtle.introduce())
print('Class default:', RegisteredPokemon.region)
```

## 4. Inheritance
A child class reuses a parent class’s attributes and methods. An electric Pokémon is a Pokémon, so it can inherit the basic setup and add an electric move.

```python
class ElectricPokemon(Pokemon):
    def use_thunderbolt(self):
        return f'{self.name} uses Thunderbolt!'

electric_pikachu = ElectricPokemon('Pikachu', 'Electric', 15)
print(electric_pikachu.name)
print(electric_pikachu.use_thunderbolt())
print(isinstance(electric_pikachu, Pokemon))
```

## 5. Multiple Inheritance
Multiple inheritance means having more than one parent class. Small helper classes called mixins add abilities such as swimming or flying. Here, `Pokemon` supplies the name.

```python
class SwimmingMixin:
    def swim(self):
        return f'{self.name} swims across the lake.'

class FlyingMixin:
    def fly(self):
        return f'{self.name} flies above the lake.'

class TravelingDragonite(SwimmingMixin, FlyingMixin, Pokemon):
    pass

dragonite = TravelingDragonite('Dragonite', 'Dragon/Flying', 55)
print(dragonite.swim())
print(dragonite.fly())
print(dragonite.name)
```

### Method lookup
Python uses the method resolution order (MRO) to decide where to look for a method. `ClassName.__mro__` shows that order. Calling a method that the object does not have raises `AttributeError`.

```python
print([parent.__name__ for parent in TravelingDragonite.__mro__])
print('Can the original Pikachu fly?', hasattr(pikachu, 'fly'))
```

## 6. Multilevel Inheritance
Multilevel inheritance forms a chain: `Pokemon` → `FirePokemon` → `TrainedFirePokemon`. The final class gets methods from both earlier classes. This models more specific classes, rather than Pokémon evolution.

```python
class FirePokemon(Pokemon):
    def use_ember(self):
        return f'{self.name} uses Ember!'

class TrainedFirePokemon(FirePokemon):
    def follow_command(self):
        return f'{self.name} follows the trainer’s command.'

trained_charmander = TrainedFirePokemon('Charmander', 'Fire', 14)
print(trained_charmander.name)
print(trained_charmander.use_ember())
print(trained_charmander.follow_command())
```

## 7. Abstract Classes
An abstract class sets required methods for its children. Use `ABC` and `@abstractmethod`. You can create a child object only after all required methods are implemented. The examples use `pass` as a placeholder; abstract methods can also contain code.

```python
from abc import ABC, abstractmethod

class BattlePokemon(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def attack(self):
        pass

    @abstractmethod
    def defend(self):
        pass

class IncompleteBattlePokemon(BattlePokemon):
    def attack(self):
        return f'{self.name} attacks!'

for pokemon_class in (BattlePokemon, IncompleteBattlePokemon):
    try:
        pokemon_class('Eevee')
    except TypeError:
        print(f'{pokemon_class.__name__} cannot be instantiated: abstract methods remain.')
```

### Implementing the full interface
This child class implements both required methods, so you can create and use an object from it.

```python
class BattleSquirtle(BattlePokemon):
    def attack(self):
        return f'{self.name} uses Water Gun!'

    def defend(self):
        return f'{self.name} withdraws into its shell.'

battle_squirtle = BattleSquirtle('Squirtle')
print(battle_squirtle.attack())
print(battle_squirtle.defend())
```

## 8. `super()` and Method Overriding
Overriding means giving a child class its own version of a method. `super()` calls the next version in the MRO, usually the parent’s. Use it to reuse setup or add to an existing method.

```python
class DescribedPokemon(Pokemon):
    def describe(self):
        return f'{self.name}: {self.pokemon_type}, level {self.level}.'

class PartnerPikachu(DescribedPokemon):
    def __init__(self, level, trainer):
        super().__init__('Pikachu', 'Electric', level)
        self.trainer = trainer

    def describe(self):
        base_description = super().describe()
        return f'{base_description} Trainer: {self.trainer}.'

partner_pikachu = PartnerPikachu(level=20, trainer='Ash')
print(partner_pikachu.describe())
```

### Replacing inherited behavior
An override can replace the old method completely. Use `super()` when you want to keep the old behavior and add more.

```python
class QuietPikachu(PartnerPikachu):
    def describe(self):
        return f'{self.name} quietly waits beside {self.trainer}.'

quiet_pikachu = QuietPikachu(level=20, trainer='Ash')
print(quiet_pikachu.describe())
```

## 9. Polymorphism and Duck Typing
Polymorphism means different objects can respond to the same method call. The loop calls `attack()` on each Pokémon, and each uses its own version.

```python
class BattleCharmander(BattlePokemon):
    def attack(self):
        return f'{self.name} uses Ember!'

    def defend(self):
        return f'{self.name} braces for the next move.'

team = [BattleSquirtle('Squirtle'), BattleCharmander('Charmander')]
for member in team:
    print(member.attack())
```

### Duck typing
Duck typing means an object can be used if it has the method you need. The simulator is not a `BattlePokemon`, but it has `attack()`, so the function can use it.

```python
class PokemonTrainingSimulator:
    def attack(self):
        return 'The simulator demonstrates a Pokémon attack.'

def demonstrate_attack(attacker):
    print(attacker.attack())

demonstrate_attack(battle_squirtle)
demonstrate_attack(PokemonTrainingSimulator())
```
