## 1. Setup and Series
A Series is one column of values with labels. Create it from a list or dictionary. `.loc` selects by label; `.iloc` selects by position, starting at zero.

```python
import pandas as pd
from io import StringIO

series = pd.Series([100, 200, 300, 400, 500], index=['a', 'b', 'c', 'd', 'e'])
print(series.loc['c'])
print(series.iloc[2])
```

```python
calories = pd.Series({'Day1': 1750, 'Day2': 2100, 'Day3': 1750, 'Day4': 2000, 'Day5': 1800})
calories.loc['Day2'] += 500
calories.iloc[3] -= 200
print(calories)
print('Above 2000:', calories[calories > 2000])
```

## 2. Creating DataFrames
A DataFrame is a table with rows and columns. Each column can have a different type. Column lists must have the same length. Put a DataFrame last in a Jupyter cell to display it as a table.

```python
team_data = {
    'Name': ['Pikachu', 'Bulbasaur', 'Squirtle'],
    'Level': [12, 10, 11],
}
team = pd.DataFrame(team_data, index=['a', 'b', 'c'])
team
```

## 3. Adding and Updating Data
Assign a new column to the DataFrame. Changing the original dictionary does not add a column to an existing DataFrame. Use `.loc[row, column]` to update a cell and `pd.concat()` to add rows.

```python
team['Role'] = ['Attacker', 'Support', 'Defender']
team.loc['a', 'Level'] = 13
new_member = pd.DataFrame({'Name': ['Charmander'], 'Level': [9], 'Role': ['Attacker']}, index=['d'])
team = pd.concat([team, new_member])
team
```

## 4. Reading CSV Data
`pd.read_csv()` reads CSV data. `index_col` chooses the row labels. This example includes CSV text. To use a file, try `pd.read_csv('pokemon.csv', index_col='No')`.

```python
csv_data = """No,Name,Type1,Type2,Height,Weight,Legendary
1,Bulbasaur,Grass,Poison,0.7,7.0,False
6,Charizard,Fire,Flying,1.7,90.0,False
9,Blastoise,Water,,1.6,85.0,False
25,Pikachu,Electric,,0.4,6.0,False
130,Gyarados,Water,Flying,6.5,235.0,False
144,Articuno,Ice,Flying,1.7,55.0,True
"""
pokemon_df = pd.read_csv(StringIO(csv_data), index_col='No')
pokemon_df
```

## 5. Inspecting and Selecting Data
Use `head(n)` to preview rows. `shape` shows row and column counts. `columns` lists names, `dtypes` shows types, and `info()` prints a table summary. `to_string(index=False)` prints without row labels.

```python
print(pokemon_df.head(3))
print('Shape:', pokemon_df.shape)
print('Columns:', pokemon_df.columns.tolist())
print(pokemon_df.dtypes)
pokemon_df.info()
```

### Selecting columns and rows
One column name returns a Series; a list of names returns a DataFrame. `set_index()` chooses new row labels. A `.loc` slice includes its ending label; an `.iloc` slice excludes its ending position.

```python
print(pokemon_df['Name'].to_string(index=False))
print(pokemon_df[['Name', 'Type1']])
pokemon_by_name = pokemon_df.set_index('Name')
print(pokemon_by_name.loc['Pikachu'])
print(pokemon_by_name.iloc[0:6:2, 0:2])
print(pokemon_by_name.loc['Charizard':'Blastoise', ['Height', 'Type1']])
```

### Handling a missing label
A missing label raises `KeyError`. Change `pokemon_name` below to try another name. The error handler prints a helpful message if the name is missing.

```python
pokemon_name = 'Pikachu'
try:
    print(pokemon_by_name.loc[pokemon_name])
except KeyError:
    print(f'{pokemon_name} is not found in the dataset.')
```

## 6. Filtering Rows
Filter rows with a condition inside `df[...]`. Combine conditions with `&` (AND), `|` (OR), or `~` (NOT). Put each comparison in parentheses.

```python
grass_pokemon = pokemon_df[pokemon_df['Type1'] == 'Grass']
tall_pokemon = pokemon_df[pokemon_df['Height'] > 2.0]
heavy_pokemon = pokemon_df[pokemon_df['Weight'] > 100.0]
print('Grass:', grass_pokemon)
print('Tall:', tall_pokemon)
print('Heavy:', heavy_pokemon)
```

```python
water_flying = pokemon_df[(pokemon_df['Type1'] == 'Water') & (pokemon_df['Type2'] == 'Flying')]
print(water_flying)
print(pokemon_df[(pokemon_df['Type1'] == 'Grass') | (pokemon_df['Type1'] == 'Electric')])
```

## 7. Aggregations
Use `mean()`, `min()`, `max()`, and `sum()` to summarize numbers. `count()` counts filled values; `len(df)` counts rows. `describe()` gives a numeric summary.

```python
measurements = pokemon_df[['Height', 'Weight']]
print('Means:', measurements.mean())
print('Minimums:', measurements.min())
print('Maximums:', measurements.max())
print('Non-missing counts:', pokemon_df.count())
print('Rows:', len(pokemon_df))
print(pokemon_df.describe())
```

### Aggregating one column
Select a column with brackets, such as `df['Height']`. This also works when a column name contains spaces or matches a DataFrame method.

```python
print('Mean height:', pokemon_df['Height'].mean())
print('Total weight:', pokemon_df['Weight'].sum())
print('Recorded primary types:', pokemon_df['Type1'].count())
```

## 8. Grouping Data
`groupby()` collects rows with the same group name. Then calculate a summary for each group. `size()` counts rows; `count()` counts filled values in a column.

```python
grouped = pokemon_df.groupby('Type1')
print('Mean height:', grouped['Height'].mean())
print('Total weight:', grouped['Weight'].sum())
print('Group sizes:', grouped.size())
summary = grouped.agg(Count=('Name', 'count'), MeanHeight=('Height', 'mean'), TotalWeight=('Weight', 'sum'))
summary
```

## 9. Cleaning and Exporting Data
Make a copy before cleaning. A missing secondary Pokémon type may mean there is no second type. Choose whether to fill or drop it. Assign cleaning results to a variable to keep them.

```python
cleaned = pokemon_df.copy()
print('Missing values:', cleaned.isna().sum())
without_legendary = cleaned.drop(columns=['Legendary'])
dual_type_only = cleaned.dropna(subset=['Type2'])
filled = cleaned.fillna({'Type2': 'None'})
print(without_legendary.head(2))
print(dual_type_only[['Name', 'Type2']])
print(filled[['Name', 'Type2']])
```

### Standardizing text and types
Use `.str` methods to clean text and `replace()` to fix labels. Use `pd.to_numeric()` to convert numeric text. The example adds inconsistent labels, then fixes them.

```python
cleaned.loc[1, 'Type1'] = ' GRASS '
cleaned.loc[6, 'Type1'] = 'Flame'
cleaned['Type1'] = cleaned['Type1'].str.strip().str.capitalize()
cleaned['Type1'] = cleaned['Type1'].replace({'Flame': 'Fire'})
cleaned['Type2'] = cleaned['Type2'].fillna('None').str.strip().str.lower()
cleaned['Height'] = pd.to_numeric(cleaned['Height'])
cleaned['Legendary'] = cleaned['Legendary'].astype('boolean')
cleaned
```

**True/False text:** map `'True'` and `'False'` to Boolean values first. Avoid `.astype(bool)` on these strings: both are nonempty, so both become `True`. The CSV example already contains Boolean values.

```python
flag_text = pd.Series(['True', 'False'])
print(flag_text.map({'True': True, 'False': False}).astype('boolean'))
```

### Removing duplicates
`drop_duplicates()` removes repeated rows. This example adds a duplicate, then removes it by name. Check that repeated names are really duplicates before deleting them.

```python
with_duplicate = pd.concat([cleaned, cleaned.iloc[[0]]])
cleaned = with_duplicate.drop_duplicates(subset=['Name'], keep='first')
print('Before:', len(with_duplicate), 'After:', len(cleaned))
```

### Exporting CSV
`to_csv()` without a path returns CSV text. To save a file, use `cleaned.to_csv('pokemon_cleaned.csv', index=True, index_label='No')`. Use `index=False` to leave out row labels.

```python
csv_output = cleaned.to_csv(index=True, index_label='No')
print(csv_output)

# Uncomment to save the file:
# cleaned.to_csv('pokemon_cleaned.csv', index=True, index_label='No')
```
