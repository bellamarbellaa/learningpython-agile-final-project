### Importing modules
A module contains reusable code. `import` makes it available; it does not install it. `math` is included with Python.

Use `import math` for names such as `math.pi`, or `from math import sqrt` to use `sqrt` directly.

## 2. Strings
Strings are text inside matching quotes. Call a method with `object.method()`. Strings cannot be changed in place; methods return new strings.

```python
character_name = 'SpongeBob SquarePants'
print(character_name.upper())
print(character_name.lower())
print(character_name.replace('SquarePants', 'the Fry Cook'))
print(character_name.split(sep=' ', maxsplit=1))
print(character_name)  # Original text is unchanged
```

Use `+` to join strings, `*` to repeat them, and `len()` to count characters. Python uses special methods such as `__add__` and `__len__` behind these operations. Use `replace()` to remove text; strings do not support subtraction.

```python
print('SpongeBob' + ' and Patrick')
print('Gary! ' * 3)
print(len('Squidward'))
print('SpongeBob SquarePants'.replace(' SquarePants', ''))
```

## 3. Arithmetic Operators
| Operator | Meaning |
|---|---|
| `+` | Add |
| `-` | Subtract |
| `*` | Multiply |
| `/` | Divide |
| `//` | Floor division (round down) |
| `%` | Remainder |
| `**` | Power |

```python
spongebob_patties = 17
patrick_patties = 5
print('Together:', spongebob_patties + patrick_patties)
print('Difference:', spongebob_patties - patrick_patties)
print('Double batch:', spongebob_patties * 2)
print('Division:', spongebob_patties / patrick_patties)
print('Full groups:', spongebob_patties // patrick_patties)
print('Left over:', spongebob_patties % patrick_patties)
print('Five squared:', patrick_patties ** 2)
```

Use parentheses to group calculations. Powers come first, then multiplication and division, then addition and subtraction. At the same level, work left to right; chained powers group right to left. `1_000` and `1000` mean the same number.

```python
spongebob_jellyfish = 10
patrick_jellyfish = 6
average_catch = (spongebob_jellyfish + patrick_jellyfish) / 2
print('Average catch:', average_catch)
print(2 + 3 * 4)
print((2 + 3) * 4)
```

## 4. Assignment Operators
`=` assigns a value. For numbers, `x += 1` means `x = x + 1`. Other updates include `-=`, `*=`, `/=`, `//=`, `%=`, and `**=`.

```python
spongebob_patties = 10
spongebob_patties += 5
spongebob_patties -= 2
print('Patties remaining:', spongebob_patties)

spongebob_score = 8
patrick_score = 3
spongebob_score, patrick_score = patrick_score, spongebob_score
print('Swapped scores:', spongebob_score, patrick_score)
```

## 5. Comparison Operators
Comparisons return `True` or `False`. Use `==` to compare and `=` to assign. `1 = 1` is invalid syntax.

| Operator | Meaning |
|---|---|
| `==` | Equal |
| `!=` | Not equal |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal |
| `<=` | Less than or equal |

```python
spongebob_jellyfish = 10
patrick_jellyfish = 6
print('Same catch:', spongebob_jellyfish == patrick_jellyfish)
print('Different catch:', spongebob_jellyfish != patrick_jellyfish)
print('SpongeBob caught more:', spongebob_jellyfish > patrick_jellyfish)
print('Patrick caught fewer:', patrick_jellyfish < spongebob_jellyfish)
print('At least ten:', spongebob_jellyfish >= 10)
print('At most six:', patrick_jellyfish <= 6)
```

## 6. Lists
Lists store ordered items inside `[]`. They allow duplicates and can be changed. Indexing starts at `0`; `-1` means the last item.

```python
characters = ['SpongeBob', 'Patrick', 'Squidward']
print(characters[0])
print(characters[-1])
characters[2] = 'Sandy'
print(characters)
```

### Add, remove, and count
`append()` adds an item at the end. `insert(index, item)` adds at a position. `remove(value)` removes the first match. `len()` counts items.

```python
characters.append('Gary')
characters.insert(2, 'Squidward')
characters.remove('Gary')
print(characters)
print('Character count:', len(characters))
```

### Combine lists and join text
`+` combines lists. `extend()` adds items from another list. `append(other_list)` adds that list as one nested item. Use `join()` to turn a list of strings into text.

```python
friends = ['SpongeBob', 'Patrick']
neighbors = ['Squidward', 'Sandy']
print(friends + neighbors)
friends.extend(neighbors)
print(friends)
print('Bikini Bottom friends: ' + ', '.join(friends))

visitors = ['Gary']
for character in friends:
    visitors.append(character)
print(visitors)
```

### Slicing and unpacking
`items[start:stop:step]` selects part of a list; `stop` is excluded. The default step is `1`. Unpacking assigns items to variables, and a starred variable collects the remaining items into a list.

```python
characters = ['SpongeBob', 'Patrick', 'Squidward', 'Sandy', 'Gary']
print(characters[:2])   # First two
print(characters[2:])   # From index 2 to the end
print(characters[-1:])  # Last item as a list
print(characters[::2])  # Every second item
print(characters[:])   # Shallow copy

spongebob, patrick = characters[:2]
first_character, *other_characters = characters
print(spongebob, patrick)
print(first_character, other_characters)
```

## 7. Dictionaries
Dictionaries store `key: value` pairs inside `{}`. Access values by key rather than numeric position. Keys are unique; values may repeat. Dictionaries can be changed and preserve insertion order.

| | List | Dictionary |
|---|---|---|
| Example | `['SpongeBob', 'Patrick']` | `{'name': 'SpongeBob'}` |
| Access | Index: `characters[0]` | Key: `character['name']` |
| Best for | A sequence of items | Named details about an item |

```python
character = {
    'name': 'SpongeBob',
    'job': 'Fry cook',
    'pet': 'Gary',
    'patties_made': 12,
}
print(character['name'])
print(character.get('hobby', 'Not recorded'))  # Default for a missing key
character['patties_made'] += 3
character['hobby'] = 'Jellyfishing'
print(character)
```

### Inspect and remove entries
`keys()` gives keys, `values()` gives values, and `items()` gives both. `pop(key)` removes and returns a value. A missing key raises `KeyError`; use `get()` when you want a default.

```python
print(list(character.keys()))
print(list(character.values()))
for key, value in character.items():
    print(key, ':', value)

removed_hobby = character.pop('hobby')
print('Removed:', removed_hobby)
print('Number of entries:', len(character))
```

### A list of dictionaries
Use a list for multiple characters, with a dictionary holding each character’s details.

```python
bikini_bottom = [
    {'name': 'SpongeBob', 'hobby': 'Jellyfishing'},
    {'name': 'Patrick', 'hobby': 'Jellyfishing'},
    {'name': 'Squidward', 'hobby': 'Clarinet'},
]
print(bikini_bottom[0]['name'])
for resident in bikini_bottom:
    print(resident['name'] + ': ' + resident['hobby'])
```
